```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Import dataset
df = pd.read_csv("Retail_Sales_Data.csv")

# 2. Explore dataset
print(df.head())
print(df.info())
print(df.describe())

# 3. Identify relevant columns
print(df[['Region', 'SalesAmount', 'ProductCategory']].head())

# 4. Group by region and calculate total sales
region_sales = df.groupby('Region')['SalesAmount'].sum()
print("\nTotal Sales by Region:")
print(region_sales)

# 5. Visualize sales by region
region_sales.plot(kind='bar', title='Sales by Region')
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()

region_sales.plot(kind='pie', autopct='%1.1f%%', title='Sales Distribution')
plt.ylabel("")
plt.show()

# 6. Find top-performing region
print("Top Performing Region:", region_sales.idxmax())

# 7. Group by region and product category
category_sales = df.groupby(
    ['Region', 'ProductCategory']
)['SalesAmount'].sum().unstack()

print("\nSales by Region and Product Category:")
print(category_sales)

# 8. Visualize region and product category sales

# Grouped bar chart
category_sales.plot(kind='bar', title='Region vs Product Category')
plt.ylabel("Total Sales")
plt.show()

# Stacked bar chart
category_sales.plot(kind='bar', stacked=True, title='Stacked Sales')
plt.ylabel("Total Sales")
plt.show()
```
