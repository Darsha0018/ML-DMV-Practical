```python
# Practical 10: Real Estate Data Wrangling

import pandas as pd

# Step 1: Load Dataset
df = pd.read_csv("RealEstate_Prices.csv")

# Step 2: Clean Column Names
df.columns = (
    df.columns.str.lower()
    .str.replace(r'[^a-z0-9]+', '_', regex=True)
    .str.strip('_')
)

# Rename columns to simple names
df.rename(columns={
    'area_sqft': 'area_sqft',
    'property_type': 'property_type',
    'sale_price': 'sale_price'
}, inplace=True)

print("Columns:", df.columns.tolist())

# Step 3: Handle Missing Values
df['area_sqft'] = df['area_sqft'].fillna(df['area_sqft'].median())
df['sale_price'] = df['sale_price'].fillna(df['sale_price'].median())

# Step 4: Convert Date
df['sale_date'] = pd.to_datetime(df['sale_date'], errors='coerce')

# Step 5: Remove Missing Dates
df.dropna(subset=['sale_date'], inplace=True)

# Step 6: Filter Houses Sold Since 2023
sub = df[
    (df['sale_date'] >= '2023-01-01') &
    (df['property_type'] == 'House')
]

print("\nHouses Sold Since 2023:", len(sub))

# Step 7: Encode Property Type
df = pd.get_dummies(df, columns=['property_type'])

# Step 8: Average Price by Neighborhood
print("\nAverage Price by Neighborhood:")
print(df.groupby('neighborhood')['sale_price'].mean())

# Step 9: Remove Outliers using IQR
q1 = df['sale_price'].quantile(0.25)
q3 = df['sale_price'].quantile(0.75)

iqr = q3 - q1

df = df[
    df['sale_price'].between(
        q1 - 1.5 * iqr,
        q3 + 1.5 * iqr
    )
]

print("\nRows After Outlier Removal:", len(df))

# Step 10: Save Cleaned Dataset
df.to_csv("cleaned_realestate.csv", index=False)

print("\nData Wrangling Completed Successfully!")
```
