```python
# Practical 7: Data Loading, Cleaning and Analysis

# Step 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

# Step 2: Load CSV, Excel and JSON
csv = pd.read_csv("data/sales.csv")
excel = pd.read_excel("data/sales.xlsx")
json = pd.read_json("data/sales.json")

# Step 3: Combine Datasets
df = pd.concat([csv, excel, json], ignore_index=True)

# Step 4: Remove Duplicate Records
df.drop_duplicates(inplace=True)

# Step 5: Check Dataset
print("Dataset:")
print(df.head())

print("\nShape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())

# Step 6: Handle Missing Values
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Price'] = df['Price'].fillna(df['Price'].median())

# Step 7: Calculate Sales
df['Sales'] = df['Quantity'] * df['Price']

# Step 8: Data Analysis
print("\nTotal Sales:", df['Sales'].sum())

print("\nAverage Sales:", df['Sales'].mean())

print("\nSales by Category:")
print(df.groupby('Category')['Sales'].sum())

print("\nSales by Region:")
print(df.groupby('Region')['Sales'].sum())

# Step 9: Visualization

# Bar Chart
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()

# Pie Chart
df.groupby('Category')['Sales'].sum().plot(
    kind='pie', autopct='%1.1f%%'
)
plt.title("Sales Distribution")
plt.ylabel("")
plt.show()

# Box Plot
df.boxplot(column='Sales', by='Category')
plt.title("Sales Distribution by Category")
plt.suptitle("")
plt.show()

# Step 10: Save Dataset
df.to_csv("cleaned_sales.csv", index=False)

print("\nPractical Completed Successfully!")
```
