```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load dataset
df = pd.read_csv("City_Air_Quality.csv")
df['Date'] = pd.to_datetime(df['Date'])

print(df.head())
print(df.describe())

# 2. AQI trend over time
plt.plot(df['Date'], df['AQI'])
plt.title("AQI Trends")
plt.xlabel("Date")
plt.ylabel("AQI")
plt.show()

# 3. Pollutants over time
for col in ['PM2.5', 'PM10', 'CO']:
    plt.plot(df['Date'], df[col], label=col)

plt.title("Pollutants Trends")
plt.xlabel("Date")
plt.ylabel("Pollutant Level")
plt.legend()
plt.show()

# 4. Monthly average pollutants
df['Month'] = df['Date'].dt.month
monthly = df.groupby('Month')[['PM2.5', 'PM10']].mean()

monthly.plot(kind='bar')
plt.title("Monthly Average Pollutants")
plt.xlabel("Month")
plt.ylabel("Average Level")
plt.show()

# 5. Box plot
df[['PM2.5', 'PM10', 'CO']].plot(kind='box')
plt.title("Pollutant Distribution")
plt.show()

# 6. Scatter plot
plt.scatter(df['PM2.5'], df['AQI'])
plt.title("PM2.5 vs AQI")
plt.xlabel("PM2.5")
plt.ylabel("AQI")
plt.show()
```
