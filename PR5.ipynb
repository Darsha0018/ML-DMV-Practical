```python
# Practical 5B: Ensemble Learning on Iris Dataset

# Step 1: Import Libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier, VotingClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score

# Step 2: Load Dataset
df = pd.read_csv("Iris.csv")

X = df.drop(columns=['species'])
y = df['species']

# Step 3: Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1
)

# Step 4: Create Models
ada = AdaBoostClassifier(n_estimators=100, random_state=1)
gbm = GradientBoostingClassifier(random_state=1)
xgb = XGBClassifier(eval_metric='mlogloss', random_state=1)

# Step 5: Voting Models
hard = VotingClassifier(
    estimators=[('a', ada), ('g', gbm), ('x', xgb)],
    voting='hard'
)

soft = VotingClassifier(
    estimators=[('a', ada), ('g', gbm), ('x', xgb)],
    voting='soft'
)

# Step 6: Train and Test All Models
models = {
    'AdaBoost': ada,
    'GBM': gbm,
    'XGBoost': xgb,
    'Hard Voting': hard,
    'Soft Voting': soft
}

for name, model in models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    print(name, "Accuracy:", round(accuracy, 3))
```
